### <u>**Create all tables**</u>

```python
# Drop tables if they already exist for fresh start
mycursor.execute("DROP TABLE Room_amenity")
mycursor.execute("DROP TABLE Booking")
mycursor.execute("DROP TABLE Room")
mycursor.execute("DROP TABLE Review")
mycursor.execute("DROP TABLE Guest")
mycursor.execute("DROP TABLE Hotel")
mycursor.execute("DROP TABLE Room_Type")
mycursor.execute("DROP TABLE Amenity")
print("Successfully dropped previous tables")

# Create new tables
mycursor.execute("""CREATE TABLE Guest (
                    user_ID        BIGINT AUTO_INCREMENT PRIMARY KEY,
                    email          VARCHAR(255) NOT NULL UNIQUE,
                    password_hash  VARCHAR(255) NOT NULL,
                    full_name      VARCHAR(200) NOT NULL,
                    phone          VARCHAR(30),
                    created_at     DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP
                );""")
print("Created Guest table")

mycursor.execute("""CREATE TABLE Hotel (
                    hotel_ID   BIGINT AUTO_INCREMENT PRIMARY KEY,
                    name       VARCHAR(200) NOT NULL,
                    address    VARCHAR(300) NOT NULL,
                    city       VARCHAR(100) NOT NULL,
                    state      VARCHAR(100),
                    country    VARCHAR(100) NOT NULL
                );""")
print("Created Hotel table")

mycursor.execute("""CREATE TABLE Room_Type (
                    room_type_ID  BIGINT AUTO_INCREMENT PRIMARY KEY,
                    capacity      INT NOT NULL,
                    name          VARCHAR(100) NOT NULL,
                    price_per_night   DECIMAL(10,2) NOT NULL
                );""")
print("Created Room_Type table")

mycursor.execute("""CREATE TABLE Room (
                    room_No       INT NOT NULL,
                    hotel_ID      BIGINT NOT NULL,
                    room_type_ID  BIGINT NOT NULL,
                    description   TEXT,
                    PRIMARY KEY (room_No, hotel_ID),
                    FOREIGN KEY (hotel_ID) REFERENCES Hotel(hotel_ID),
                    FOREIGN KEY (room_type_ID) REFERENCES Room_Type(room_type_ID)
                );""")
print("Created Room table")

mycursor.execute("""CREATE TABLE Booking (
                    Booking_ID     BIGINT AUTO_INCREMENT PRIMARY KEY,
                    user_ID        BIGINT NOT NULL,
                    room_No        INT NOT NULL,
                    hotel_ID       BIGINT NOT NULL,
                    check_in_date  DATE NOT NULL,
                    check_out_date DATE NOT NULL,
                    status         VARCHAR(20) NOT NULL,
                    created_at     DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
                    FOREIGN KEY (user_ID) REFERENCES Guest(user_ID),
                    FOREIGN KEY (room_No, hotel_ID) REFERENCES Room(room_No, hotel_ID),
                    FOREIGN KEY (hotel_ID) REFERENCES Hotel(hotel_ID)
                );""")
print("Created Booking table")

mycursor.execute("""CREATE TABLE Review (
                    review_id   BIGINT AUTO_INCREMENT PRIMARY KEY,
                    user_ID     BIGINT NOT NULL,
                    hotel_ID    BIGINT NOT NULL,
                    rating      INT NOT NULL,
                    title       VARCHAR(200),
                    body        TEXT,
                    created_at  DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
                    FOREIGN KEY (user_ID) REFERENCES Guest(user_ID),
                    FOREIGN KEY (hotel_ID) REFERENCES Hotel(hotel_ID)
                );""")
print("Created Review table")

mycursor.execute("""CREATE TABLE Amenity (
                    amenity_ID    BIGINT AUTO_INCREMENT PRIMARY KEY,
                    name          VARCHAR(100) NOT NULL,
                    description   TEXT,
                    UNIQUE (name)
                );""")
print("Created Amenity table")

mycursor.execute("""CREATE TABLE Room_amenity (
                    amenity_ID  BIGINT NOT NULL,
                    room_No     INT NOT NULL,
                    hotel_ID    BIGINT NOT NULL,
                    PRIMARY KEY (amenity_ID, room_No, hotel_ID),
                    FOREIGN KEY (amenity_ID) REFERENCES Amenity(amenity_ID),
                    FOREIGN KEY (room_No) REFERENCES Room(room_No),
                    FOREIGN KEY (hotel_ID) REFERENCES Hotel(hotel_ID)
                );""")
print("Created Room_amenity table")
```

### <u>**5 Queries from Question 4 of Milestone 2**</u>

#### Q4.1

```python
# List each booking with guest name and email, hotel and location,
# room number and type, check in and out dates, and booking status

mycursor.execute("""SELECT
                    b.booking_id,
                    g.full_name,
                    g.email,
                    h.name AS hotel_name,
                    h.city,
                    h.state,
                    b.hotel_id,
                    b.room_no,
                    rt.name AS room_type,
                    rt.capacity,
                    b.check_in_date,
                    b.check_out_date,
                    b.status,
                    b.created_at
                FROM Booking b
                JOIN Guest g
                ON b.user_id = g.user_id
                JOIN Hotel h
                ON b.hotel_id = h.hotel_id
                JOIN Room r
                ON b.hotel_id = r.hotel_id
                AND b.room_no  = r.room_no
                JOIN Room_Type rt
                ON r.room_type_id = rt.room_type_id
                ORDER BY b.created_at DESC;""")

myresult = mycursor.fetchall()

for x in myresult:
  print(x)
```

#### Q4.2

```python
# This SQL statement finds hotels 
# whose average review rating is higher than
# the overall average rating across all reviews.

mycursor.execute("""SELECT
                	h.hotel_ID,
                	h.name AS hotel_name,
                	AVG(r.rating) AS hotel_avg_rating
                FROM Hotel h
                JOIN Review r
                ON r.hotel_ID = h.hotel_ID
                GROUP BY h.hotel_ID, h.name
                HAVING AVG(r.rating) > (
                	SELECT AVG(rating)
                	FROM Review
                );""")

myresult = mycursor.fetchall()

for x in myresult:
  print(x)
```

#### Q4.3

```python
# This query displays all hotels available in the
# hotel database that do not yet have any reviews.

mycursor.execute("""SELECT hotel_ID
                FROM Hotel
                WHERE hotel_ID NOT IN (SELECT hotel_ID FROM Review);""")

myresult = mycursor.fetchall()

for x in myresult:
    print(x)
```

#### Q4.4

```python
# This SQL statement finds the average review score from users
# who created their account within the past year

mycursor.execute("""SELECT
                	g.user_id,
                	g.full_name,
                	AVG(r.rating) AS avg_rating
                FROM Guest g
                JOIN Review r
                ON r.user_id = g.user_id
                WHERE g.created_at >= date_sub(now(), interval 1 year)
                GROUP BY g.user_id, g.full_name
                ORDER BY avg_rating;""")

myresult = mycursor.fetchall()

for x in myresult:
    print(x)
```

#### Q4.5

```python
# This SQL statement is a left outer join example
# It lists all hotels and shows booking info if
# they have bookings hotels with no bookings will
# still appear (booking columns will be NULL)

mycursor.execute("""SELECT
                	h.hotel_id,
                	h.name as hotel_name,
                	h.city,
                	h.state,
                	b.booking_id,
                	b.status,
                	b.check_in_date,
                	b.check_out_date
                FROM Hotel h
                LEFT JOIN Booking b
                	ON b.hotel_id = h.hotel_id
                ORDER BY h.hotel_id, b.created_at;""")

myresult = mycursor.fetchall()

for x in myresult:
    print(x)
```
